```python
import pandas as pd

# LISTA DE DADOS
data = {
    'Nome': ['Ben', 'Gwen', 'Max', 'Flanders'],
    'Idade': [20, 25, 48, 33],
    'Saldo': [1800, 650, 5500 , 4500]
}

# CREATE FILE .CSV

df = pd.DataFrame(data)
df.to_csv('clientes.csv', index=False)

# EXTRACT

df_clientes = pd.read_csv('clientes.csv')
print("Dados Extraidos: ")
print(df_clientes)

# TRANSFORM

def mensagem_padrao_ia (nome, saldo):
  if saldo > 4000:
    return f"Olá {nome}, vimos seu excelente saldo e te recomendamos fortemente a investir em Tesouro Direto!"
  else:
    return f"Olá {nome}, poupar é o primeiro passo para investir. Vamos começar hoje?"

df_clientes['Mensagem'] = df_clientes.apply(lambda x: mensagem_padrao_ia(x['Nome'], x['Saldo']), axis=1)

print("\nDados Transformados:")
print(df_clientes)

# LOAD

df_clientes.to_csv('clientes_com_mensagens.csv', index=False)
print("\nArquivo 'clientes_com_mensagens.csv' gerado com sucesso!")
```
